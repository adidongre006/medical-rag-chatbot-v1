### 1. Groq API test 

```python
from dotenv import load_dotenv
import os
from langchain_groq import ChatGroq

load_dotenv()

llm = ChatGroq(
    api_key=os.environ.get("Groq_API_KEY"),
    model="openai/gpt-oss-120b",
    temperature=0.7,
   )
response = llm.invoke("What is the capital of France?")

print(response.content)
```
